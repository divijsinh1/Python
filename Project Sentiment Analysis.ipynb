```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from textblob import TextBlob  #natural sentence / language meaning understnanding
from wordcloud import WordCloud #Creating the wordcloud
```

```python
#load the file 

data = pd.read_csv("Sentiment_Analysis .csv")

data.head(10)
```

```python
# define the sentiment analysis function 

def get_sentiment(text):

    analysis = TextBlob(text)

    if analysis.sentiment.polarity >0:
        return "Positive"

    elif analysis.sentiment.polarity<0:
        return "Negative"

    else:
        return"Neutral"
```

```python
# apply sentiment analysis 

data["Sentiment"] = data["Sentences"].apply(get_sentiment)
```

```python
data.to_excel("Sentiment_upd.xlsx",index=False)
```
